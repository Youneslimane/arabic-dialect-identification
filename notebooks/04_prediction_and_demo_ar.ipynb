# المرحلة الرابعة: التنبؤ وتجربة النظام

نستخدم النماذج المحفوظة لتصنيف نصوص جديدة دون إعادة التدريب، ثم نعرض النتيجة بالعربية.


## طريقة التنبؤ

1. تنظيف النص.
2. تحويله إلى Character TF-IDF.
3. تحديد المنطقة.
4. تحميل نموذج الدولة الخاص بالمنطقة.
5. إعادة النتيجة في قاموس.


````python
# sys لإضافة مجلد scripts إلى مسار الاستيراد
import sys
# Path للتعامل مع مسارات الملفات
from pathlib import Path
# pandas لعرض البيانات والإحصاءات في جداول
import pandas as pd

# نحدد جذر المشروع الحالي
ROOT = Path.cwd()
sys.path.insert(0, str(ROOT / "scripts"))
# نستورد دالة التنبؤ الهرمي
from predict import predict
````

````python
# أمثلة نصية من لهجات مختلفة
examples = [
    "شو أخبارك اليوم؟",
    "إزيك عامل إيه؟",
    "واش خبارك اليوم؟",
]

# قائمة لتجميع نتائج التنبؤ
predictions = []
# نطبق النموذج على كل مثال
for text in examples:
# ننفذ التطبيع والتصنيف الهرمي
    result = predict(text)
    predictions.append({
        "النص": text,
        "النص بعد التطبيع": result["text"],
        "المنطقة": result["region"],
        "الدولة": result["country"],
    })

# نعرض النتائج في جدول
pd.DataFrame(predictions)
````

## تفسير النتائج

النتائج تقديرية وليست ضمانًا مطلقًا. النصوص القصيرة جدًا أو اللهجات المتقاربة قد تسبب
التباسًا. الواجهة `app.py` تستخدم الدالة نفسها، لكنها تضيف عرضًا عربيًا وشريط تقدم وصورة
علم الدولة.


````python
REGION_ARABIC = {
    "Egyptian": "المصرية",
    "Gulf": "الخليجية",
    "Iraqi": "العراقية",
    "Levantine": "الشامية",
    "Maghrebi": "المغاربية",
    "MSA": "الفصحى",
}

for item in predictions:
    print(
        f'{REGION_ARABIC.get(item["المنطقة"], item["المنطقة"])} '
        f'/ {item["الدولة"]}'
    )
````
